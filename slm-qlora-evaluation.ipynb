## Cell 4 - Fungsi Inferensi dan Paper

```python
import collections
import re
import string
```

```python
def normalize_qa_answer(s):
    """Membersihkan teks dari tanda baca, spasi berlebih, dan mengubah ke huruf kecil."""
    def remove_punc(text):
        exclude = set(string.punctuation)
        return ''.join(ch for ch in text if ch not in exclude)
    def white_space_fix(text):
        return ' '.join(text.split())
    def lower(text):
        return text.lower()
    return white_space_fix(remove_punc(lower(str(s))))

def qa_exact_match(prediction, ground_truth):
    return 1.0 if normalize_qa_answer(prediction) == normalize_qa_answer(ground_truth) else 0.0

def qa_f1_score(prediction, ground_truth):
    pred_tokens = normalize_qa_answer(prediction).split()
    truth_tokens = normalize_qa_answer(ground_truth).split()

    # Jika keduanya kosong, maka dianggap benar (match)
    if not pred_tokens and not truth_tokens:
        return 1.0
    # Jika salah satu kosong tapi yang lain tidak
    if not pred_tokens or not truth_tokens:
        return 0.0

    common = collections.Counter(pred_tokens) & collections.Counter(truth_tokens)
    num_same = sum(common.values())

    if num_same == 0:
        return 0.0

    precision = 1.0 * num_same / len(pred_tokens)
    recall = 1.0 * num_same / len(truth_tokens)
    f1 = (2 * precision * recall) / (precision + recall)
    return f1
    
def extract_hybrid_prediction(task_name, gen_text):
    """
    Mengekstrak label jawaban secara pemaaf menggunakan Regex,
    mengakomodasi model finetuned yang mengalami Excessive Verbosity.
    """
    gen_text = str(gen_text).strip().upper()

    if task_name in ["copal", "indoculture"]:
        # 1. Deteksi Jebakan "A": Cari huruf A, B, C di awal kalimat yang diikuti titik, spasi, atau kurung
        match = re.search(r'^([A-C])[\.\s\(\:]', gen_text)
        if match:
            return match.group(1)

        # 2. Jika model hanya menjawab murni 1 huruf
        if len(gen_text) == 1 and gen_text in ['A', 'B', 'C']:
            return gen_text

        # 3. Cari pola kalimat penjabaran panjang
        for label in ["A", "B", "C"]:
            if f"JAWABAN: {label}" in gen_text or f"ADALAH {label}" in gen_text or f"PILIHAN {label}" in gen_text:
                return label

    elif task_name == "indonli":
        # Sinkronisasi dengan map_hybrid_label ("SESUAI", "NETRAL", "BERTENTANGAN")
        # Menangkap variasi bahasa Inggris dari instruksi dan menerjemahkannya
        if "SESUAI" in gen_text or "ENTAILMENT" in gen_text: return "SESUAI"
        if "BERTENTANGAN" in gen_text or "CONTRADICTION" in gen_text: return "BERTENTANGAN"
        if "NETRAL" in gen_text or "NEUTRAL" in gen_text: return "NETRAL"

    elif task_name == "smsa":
        # Menangkap sentimen bahasa Indonesia dan Inggris
        if "POSITIF" in gen_text or "POSITIVE" in gen_text: return "POSITIF"
        if "NEGATIF" in gen_text or "NEGATIVE" in gen_text: return "NEGATIF"
        if "NETRAL" in gen_text or "NEUTRAL" in gen_text: return "NETRAL"

    elif task_name == "qa":
        # Ambil baris pertama, bersihkan prefix "A (" jika model terjebak format pilihan ganda
        clean_text = gen_text.split('\n')[0].strip()
        match = re.search(r'^[A-C]\s*[\(\.\:]\s*(.*)', clean_text)
        if match:
            return match.group(1).replace(")", "").strip()
        return clean_text

    return "UNKNOWN"

def map_hybrid_label(task_name, sample):
    """Menyamakan format label dataset (Mengatasi tipe data string/integer yang bentrok)"""

    # 1. QA punya cara unik sendiri (menggunakan kolom 'answers')
    if task_name == "qa":
       passage_tokens = sample.get("passage", [])
       labels = sample.get("seq_label", [])

       # Ekstrak token yang labelnya 1 (Begin) atau 2 (Inside)
       ans_tokens = [tok for tok, lbl in zip(passage_tokens, labels) if lbl in [1, 2]]

       # Gabungkan menjadi kalimat utuh
       true_answer = " ".join(ans_tokens).strip().upper()

       # Opsional: bersihkan spasi sebelum tanda baca (misal "Christian , Hadinata" jadi "Christian, Hadinata")
       true_answer = true_answer.replace(" ,", ",").replace(" .", ".")
       return true_answer if true_answer else "UNKNOWN"

    # 2. Untuk dataset selain QA
    raw_label = str(sample.get("label", sample.get("answer", sample.get("target", "-1")))).strip().upper()

    if task_name == "copal":
        # Ambil langsung dari kolom 'label'
        # 0 -> A, 1 -> B
        return "A" if raw_label == "0" else "B"

    elif task_name == "indoculture":
        label_map = {"0": "A", "1": "B", "2": "C"}
        return raw_label if raw_label in ["A", "B", "C"] else label_map.get(raw_label, "UNKNOWN")


    elif task_name == "indonli":
        # Ambil langsung dari kolom 'label'
        # raw_label = str(sample.get("label", "-1")).strip()
        # Petakan dari angka ke teks
        label_map = {"0": "SESUAI", "1": "NETRAL", "2": "BERTENTANGAN"}
        return label_map.get(raw_label, "UNKNOWN")

    elif task_name == "smsa":
        label_map = {"0": "POSITIF", "1": "NETRAL", "2": "NEGATIF"}
        return raw_label if raw_label in ["POSITIF", "NETRAL", "NEGATIF"] else label_map.get(raw_label, "UNKNOWN")

    return raw_label


def evaluate_batched(model, tokenizer, dataset, task_name, batch_size=8):
    model.eval()
    results = []

    for i in tqdm(range(0, len(dataset), batch_size), desc=f"Eval {task_name}"):
        batch = dataset.select(range(i, min(i + batch_size, len(dataset))))
        prompts = []

        for s in batch:
            # 1. Logic Prompting Spesifik per Dataset
            if task_name == "copal":
                # Baca tipe pertanyaan (cause/effect)
                q_type = str(s.get("question", "")).lower()

                if q_type == "cause":
                    pertanyaan = "Apa penyebab dari situasi tersebut?"
                elif q_type == "effect":
                    pertanyaan = "Apa akibat dari situasi tersebut?"
                else:
                    pertanyaan = "Manakah pilihan yang paling masuk akal?" # Fallback

                p = (
                    f"Situasi: {s['premise']}\n"
                    f"Pertanyaan: {pertanyaan}\n"
                    f"A. {s['choice1']}\n"
                    f"B. {s['choice2']}\n"
                    "Jawaban (A/B):"
                )
            # Tambahkan ini di dalam fungsi pembuat prompt-mu di Cell 6
            elif task_name == "qa":
                # GABUNGKAN token list menjadi string teks biasa menggunakan " ".join()
                passage_text = " ".join(s.get("passage", [])).replace(" ,", ",").replace(" .", ".")
                question_text = " ".join(s.get("question", [])).replace(" ,", ",").replace(" ?", "?")

                p = (
                    "Berdasarkan teks berikut, jawablah pertanyaan dengan singkat dan tepat.\n\n"
                    f"Teks: {passage_text}\n"
                    f"Pertanyaan: {question_text}\n\n"
                    "Jawaban:"
                )
                # Jika pakai apply_chat_template:
                # messages = [{"role": "user", "content": p}]
                # prompt = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
            elif task_name == "indoculture":
                # Ekstrak list pilihan menjadi baris-baris teks biasa
                options_list = s.get("options", [])

                # Cek jika formatnya list, gabungkan dengan enter (\n)
                if isinstance(options_list, list):
                    options_text = "\n".join(options_list)
                else:
                    options_text = str(options_list) # Fallback jika sudah string

                p = (
                    f"Konteks: {s['context']}\n"
                    f"Pilihan:\n{options_text}\n\n"
                    "Jawaban (A/B/C):"
                )
            elif task_name == "indonli":
                p = f"Premis: {s['premise']}\nHipotesis: {s['hypothesis']}\nHubungan (Entailment/Contradiction/Neutral):"
            else: # smsa
                p = f"Tentukan sentimen kalimat berikut (positif/negatif/netral):\n'{s['text']}'\nSentimen:"

            # 2. Bungkus dengan Chat Template
            msg = [{"role": "user", "content": p}]
            prompts.append(tokenizer.apply_chat_template(msg, tokenize=False, add_generation_prompt=True))

        inputs = tokenizer(prompts, return_tensors="pt", padding=True, truncation=True).to(model.device)

        with torch.no_grad():
            start_inf = time.perf_counter()
            # Gunakan max_new_tokens kecil agar cepat
            outputs = model.generate(**inputs, max_new_tokens=10, do_sample=False, pad_token_id=tokenizer.eos_token_id)
            end_inf = time.perf_counter()

        for j, out in enumerate(outputs):
            gen_text = tokenizer.decode(out[inputs["input_ids"].shape[1]:], skip_special_tokens=True).upper()
            results.append({"pred": gen_text, "latency": (end_inf - start_inf)/len(batch)})

    return results

def format_prompt(sample, task_name, condition_name):
    """
    Membangun isi konten untuk "user_msg". 
    TIDAK menggunakan ### Respons: karena add_generation_prompt=True sudah mengurusnya.
    """
    # Tentukan tag (Hanya untuk C3/C4)
    tag_prefix = ""
    if condition_name in ["C3_Linguistic", "C4_Vocabulary"]:
        # tag_prefix = "[INFORMAL]\n" if task_name == "smsa" else "[FORMAL]\n"
        tag_prefix = "[FORMAL]\n"
    
    # 1. INDONLI (Sesuai training: TIDAK ADA TAG, TIDAK ADA KATA "Instruksi:")
    if task_name == "indonli":
        p, h = sample.get("premise", "").strip(), sample.get("hypothesis", "").strip()
        # Jika C2 (tanpa tag di dataset lain), IndoNLI tetap sama bentuknya
        return (
            "Diberikan sebuah Premis dan Hipotesis. Tentukan hubungan logis di antara keduanya.\n"
            "Jawab HANYA dengan satu kata: ENTAILMENT, NEUTRAL, atau CONTRADICTION.\n\n"
            f"Premis: {p}\n"
            f"Hipotesis: {h}"
        )

    # 2. COPAL
    elif task_name == "copal":
        q_type = str(sample.get("question", "")).lower()
        pertanyaan = "Apa penyebab dari situasi tersebut?" if q_type == "cause" else \
                     "Apa akibat dari situasi tersebut?" if q_type == "effect" else \
                     "Manakah pilihan yang paling masuk akal?"
        inst_text = (
            f"Situasi: {sample['premise']}\n"
            f"Pertanyaan: {pertanyaan}\n"
            f"A. {sample['choice1']}\n"
            f"B. {sample['choice2']}\n"
            "Jawaban (A atau B):"
        )
        return f"{tag_prefix}Instruksi:\n{inst_text}" if tag_prefix else f"Instruksi:\n{inst_text}"

    # 3. INDOCULTURE
    elif task_name == "indoculture":
        options_list = sample.get("options", [])
        options_text = "\n".join(options_list) if isinstance(options_list, list) else str(options_list)
        inst_text = (
            "Bacalah pertanyaan berikut dan pilih jawaban yang paling tepat berdasarkan konteks budaya Indonesia.\n\n"
            f"Konteks: {sample['context']}\n"
            f"Pilihan:\n{options_text}\n\n"
            "Jawaban (A, B, atau C):"
        )
        return f"{tag_prefix}Instruksi:\n{inst_text}" if tag_prefix else f"Instruksi:\n{inst_text}"

    # 4. SMSA
    elif task_name == "smsa":
        text = sample.get("text", sample.get("sentence", "")).strip()
        inst_text = (
            "Analisis sentimen dari teks berikut. Jawab HANYA dengan satu kata: POSITIF, NETRAL, atau NEGATIF.\n\n"
            f"Teks: {text}\n"
            "Jawaban:"
        )
        return f"{tag_prefix}Instruksi:\n{inst_text}" if tag_prefix else f"Instruksi:\n{inst_text}"

    # 5. QA
    elif task_name == "qa":
        passage_text = " ".join(sample.get("passage", [])).replace(" ,", ",").replace(" .", ".")
        question_text = " ".join(sample.get("question", [])).replace(" ,", ",").replace(" ?", "?")
        inst_text = (
            "Berdasarkan teks berikut, jawablah pertanyaan dengan singkat dan tepat.\n\n"
            f"Teks: {passage_text}\n"
            f"Pertanyaan: {question_text}\n\n"
            "Jawaban:"
        )
        return f"{tag_prefix}Instruksi:\n{inst_text}" if tag_prefix else f"Instruksi:\n{inst_text}"
        
    return ""

def evaluate_batched_final(model, tokenizer, dataset, task_name, condition_name, batch_size=8):
    model.eval()
    results = []

    for i in tqdm(range(0, len(dataset), batch_size), desc=f"Eval {task_name} ({condition_name})"):
        batch = dataset.select(range(i, min(i + batch_size, len(dataset))))
        prompts = []

        for s in batch:
            # 1. Dapatkan string konten user
            p_content = format_prompt(s, task_name, condition_name)

            # 2. BUNGKUS DENGAN CHATML (Sama persis seperti saat training C3/C4)
            msg = [{"role": "user", "content": p_content}]
            
            # add_generation_prompt=True otomatis menambahkan <|im_start|>assistant\n
            full_prompt = tokenizer.apply_chat_template(msg, tokenize=False, add_generation_prompt=True)
            prompts.append(full_prompt)

        # --- Proses Inferensi ---
        inputs = tokenizer(prompts, return_tensors="pt", padding=True, truncation=True).to(model.device)

        with torch.no_grad():
            start_inf = time.perf_counter()
            outputs = model.generate(**inputs, max_new_tokens=10, do_sample=False, pad_token_id=tokenizer.eos_token_id)
            end_inf = time.perf_counter()

        for j, out in enumerate(outputs):
            gen_text = tokenizer.decode(out[inputs["input_ids"].shape[1]:], skip_special_tokens=True).upper()
            results.append({"pred": gen_text, "latency": (end_inf - start_inf)/len(batch)})

    return results

def evaluate_batched_c3(model, tokenizer, dataset, task_name, condition, batch_size=8):
    model.eval()
    results = []
    # "Kata sandi" wajib untuk model C3 agar tidak mengalami mismatch
    tag = "" if condition == "C2_QLoRA" else "[FORMAL]"

    for i in tqdm(range(0, len(dataset), batch_size), desc=f"Eval {task_name}"):
        batch = dataset.select(range(i, min(i + batch_size, len(dataset))))
        prompts = []

        for s in batch:
            # 1. Logic Prompting Spesifik per Dataset
            if task_name == "copal":
                q_type = str(s.get("question", "")).lower()
                pertanyaan = "Apa penyebab dari situasi tersebut?" if q_type == "cause" else \
                             "Apa akibat dari situasi tersebut?" if q_type == "effect" else \
                             "Manakah pilihan yang paling masuk akal?"
                if condition == "C2_QLoRA": 
                    p = (
                        f" ### Instruksi:\n"
                        f"Situasi: {s['premise']}\n"
                        f"Pertanyaan: {pertanyaan}\n"
                        f"A. {s['choice1']}\n"
                        f"B. {s['choice2']}\n"
                        "Jawaban (A/B):"
                    )
                else: 
                    p = (
                        f"{tag}\n### Instruksi:\n"
                        f"Situasi: {s['premise']}\n"
                        f"Pertanyaan: {pertanyaan}\n"
                        f"A. {s['choice1']}\n"
                        f"B. {s['choice2']}\n"
                        "Jawaban (A/B):"
                    )
                print(f"copal: {p}")

            elif task_name == "qa":
                passage_text = " ".join(s.get("passage", [])).replace(" ,", ",").replace(" .", ".")
                question_text = " ".join(s.get("question", [])).replace(" ,", ",").replace(" ?", "?")

                if condition == "C2_QLoRA": 
                    p = (
                        f"### Instruksi:\n"
                        "Berdasarkan teks berikut, jawablah pertanyaan dengan singkat dan tepat.\n\n"
                        f"Teks: {passage_text}\n"
                        f"Pertanyaan: {question_text}\n\n"
                        "Jawaban:"
                    )
                else:     
                    p = (
                        f"{tag}\n### Instruksi:\n"
                        "Berdasarkan teks berikut, jawablah pertanyaan dengan singkat dan tepat.\n\n"
                        f"Teks: {passage_text}\n"
                        f"Pertanyaan: {question_text}\n\n"
                        "Jawaban:"
                    )

            elif task_name == "indoculture":
                options_list = s.get("options", [])
                options_text = "\n".join(options_list) if isinstance(options_list, list) else str(options_list)

                if condition == "C2_QLoRA": 
                    p = (
                        f"### Instruksi:\n"
                        "Bacalah pertanyaan berikut dan pilih jawaban yang paling tepat hanya satu berdasarkan konteks budaya Indonesia.\n\n"
                        f"Konteks: {s['context']}\n"
                        f"Pilihan:\n{options_text}\n\n"
                        "Jawaban (A, B, atau C):"
                    )    

                else:   
                    p = (
                        f"{tag}\n### Instruksi:\n"
                        "Bacalah pertanyaan berikut dan pilih jawaban yang paling tepat hanya satu berdasarkan konteks budaya Indonesia.\n\n"
                        f"Konteks: {s['context']}\n"
                        f"Pilihan:\n{options_text}\n\n"
                        "Jawaban (A, B, atau C):"
                    )

            elif task_name == "indonli":
                # Diperbarui: Menggunakan Tag [FORMAL] agar performa kembali melonjak
                p = (
                    f"### Instruksi:\n"
                    "Diberikan sebuah Premis dan Hipotesis. Tentukan hubungan logis di antara keduanya.\n"
                    "Jawab HANYA dengan satu kata: ENTAILMENT, NEUTRAL, atau CONTRADICTION.\n\n"
                    f"Premis: {s['premise']}\n"
                    f"Hipotesis: {s['hypothesis']}\n"
                    "Jawaban:"
                )

            else: # smsa
                # Diperbarui: Menggunakan format yang memberikan akurasi 96.67% pada diagnosa
                if condition == "C2_QLoRA":
                    p = (
                        f"### Instruksi:\n"
                        "Analisis sentimen dari teks berikut.\n"
                        "Jawab HANYA dengan satu kata: POSITIF, NETRAL, atau NEGATIF.\n\n"
                        f"Teks: {s['text']}\n"
                        "Jawaban:"
                    )
                else: 
                    p = (
                        f"{tag}\n### Instruksi:\n"
                        "Analisis sentimen dari teks berikut.\n"
                        "Jawab HANYA dengan satu kata: POSITIF, NETRAL, atau NEGATIF.\n\n"
                        f"Teks: {s['text']}\n"
                        "Jawaban:"
                    )

            # 2. Bungkus dengan Chat Template (Format ChatML Qwen)
            msg = [{"role": "user", "content": p}]
            prompts.append(tokenizer.apply_chat_template(msg, tokenize=False, add_generation_prompt=True))

        # --- Sisa kode inference (inputs, outputs, decode) tetap sama seperti kodemu ---
        inputs = tokenizer(prompts, return_tensors="pt", padding=True, truncation=True).to(model.device)

        with torch.no_grad():
            start_inf = time.perf_counter()
            outputs = model.generate(**inputs, max_new_tokens=10, do_sample=False, pad_token_id=tokenizer.eos_token_id)
            end_inf = time.perf_counter()

        for j, out in enumerate(outputs):
            gen_text = tokenizer.decode(out[inputs["input_ids"].shape[1]:], skip_special_tokens=True).upper()
            results.append({"pred": gen_text, "latency": (end_inf - start_inf)/len(batch)})

    return results


```

## Cell 5 - Evaluation

```python
all_conditions_metrics = []

for condition_name, adapter_path in ADAPTERS.items():
    print(f"\n" + "="*60)
    print(f"🔥 MENGUJI KONDISI: {condition_name}")
    print("="*60)

    condition_dir = os.path.join(OUTPUT_DIR, condition_name)
    os.makedirs(condition_dir, exist_ok=True)

    # --------------------------------------------------------
    # A. MUAT MODEL & TOKENIZER
    # --------------------------------------------------------
    print(f"⏳ Memuat Tokenizer dari {adapter_path}...")
    tokenizer = AutoTokenizer.from_pretrained(adapter_path)

    tokenizer.padding_side = "left"

    print("⏳ Memuat Base Model Qwen 1.5B (4-bit)...")
    base_model = AutoModelForCausalLM.from_pretrained(
        "Qwen/Qwen2-1.5B",
        quantization_config=bnb_config,
        device_map="auto",
        trust_remote_code=True,
    )

    # 🚨 LANGKAH WAJIB: Lebarkan rak memori (Aman untuk semua kondisi)
    base_model.resize_token_embeddings(len(tokenizer))

    print("⏳ Menempelkan Adapter LoRA...")
    model = PeftModel.from_pretrained(base_model, adapter_path)
    print("✅ Model siap dievaluasi!")

    # --------------------------------------------------------
    # B. LOOP EVALUASI PER TASK
    # --------------------------------------------------------
    for task_name, dataset in eval_tasks.items():
        print(f"\n   Mengevaluasi {task_name.upper()} ({len(dataset)} sampel)...")

        # Ambil 150 sampel (sama persis dengan limit evaluasimu sebelumnya)
        # subset_ic = eval_tasks["indoculture"].select(range(min(150, len(eval_tasks["indoculture"]))))

        raw_results = evaluate_batched_final(model, tokenizer, eval_tasks[task_name], task_name=task_name, condition_name=condition_name, batch_size=1)

        y_true, y_pred = [], []
        format_errors = 0
        detailed_logs = []

        for i, res in enumerate(raw_results):
            raw_pred = res["pred"]
            clean_pred = extract_hybrid_prediction(task_name, raw_pred)
            gt_label = map_hybrid_label(task_name, dataset[i])

            y_true.append(gt_label)
            y_pred.append(clean_pred)
            if clean_pred == "UNKNOWN": format_errors += 1

            detailed_logs.append({
                "id": i,
                "text": str(dataset[i]),
                "ground_truth": gt_label,
                "prediction": clean_pred,
                "raw_output": raw_pred
            })

        # Hitung Metrik
        acc = accuracy_score(y_true, y_pred)
        unique_labels = list(set(y_true))
        macro_f1 = f1_score(y_true, y_pred, labels=unique_labels, average="macro", zero_division=0)

        print(f"      📊 Akurasi: {acc*100:.2f}% | F1: {macro_f1*100:.2f}% | Gagal Format: {format_errors}")

        all_conditions_metrics.append({
            "Condition": condition_name,
            "Task": task_name.upper(),
            "Accuracy": acc,
            "Macro_F1": macro_f1,
            "Format_Errors": format_errors
        })

        # Simpan log json per task
        with open(os.path.join(condition_dir, f"{task_name}_log.json"), "w", encoding="utf-8") as f:
            json.dump(detailed_logs, f, ensure_ascii=False, indent=2)

    # --------------------------------------------------------
    # C. CUCI OTAK (MEMBERSIHKAN VRAM GPU)
    # --------------------------------------------------------
    print(f"\n🧹 Membersihkan VRAM GPU...")
    del model
    del base_model
    del tokenizer
    gc.collect()
    torch.cuda.empty_cache()
    print("✅ VRAM bersih. Siap untuk putaran berikutnya.")
```

```python
# --------------------------------------------------------
# 5. SIMPAN HASIL AKHIR (CSV)
# --------------------------------------------------------
df_final = pd.DataFrame(all_conditions_metrics)
final_csv_path = os.path.join(OUTPUT_DIR, "METRIK_ABLATION_STUDY.csv")
df_final.to_csv(final_csv_path, index=False)

print("\n" + "="*60)
print("🎉🎉 SELURUH EVALUASI SELESAI SENGAN SUKSES! 🎉🎉")
print(f"Tabel rangkuman siap digunakan: {final_csv_path}")
print("="*60)
print(df_final.to_string(index=False))
```
